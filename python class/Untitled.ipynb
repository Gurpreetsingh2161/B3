```python
point=[(1,2),(1,3),(2,4),(5,7)]

for x, y in point:
    distance = (x**2 + y**2)** 0.5
    print(distance)
```

```python
def distance_point(x1,y1,x2,y2):
    return ((x2 - x1)**2 + (y2 - y1)**2) ** 0.5
points = [(1,2),(1,3),(2,4),(5,7)]
origin = (0,0)
for x, y in points:
    distances = distance_point(x,y,origin[0],origin[1])
    print(distances)
```
